```python
import pandas as pd
import requests
import os


url = 'https://gist.githubusercontent.com/tiangechen/b68782efa49a16edaf07dc2cdaa855ea/raw/0c794a9717f18b094eabab2cd6a6b9a226903577/movies.csv'

file_name = 'movies.csv'

if not os.path.exists(file_name):
    response = requests.get(url)
    if response.status_code == 200:
        with open(file_name, 'wb') as file:
            file.write(response.content)
        print('Successfully Downloaded!')
else:
    print('File has already been downloaded!')

df = pd.read_csv(file_name)
df = df.drop_duplicates()
```

```python
print(df.columns)
print(df.dtypes)
print(df.info)
print(df.describe())
df
```

```python
count_unique_movies = len(df['Film'].unique())
count_unique_movies
```

```python
# movies_per_year = df['Year'].value_counts()
movies_per_year = df.groupby(by='Year').count()['Film']
movies_per_year
```

```python
max_profit = df['Profitability'].max()
min_profit = df['Profitability'].min()

most_profitable = df[df['Profitability'] == max_profit]
most_profitable
print(most_profitable[['Film', 'Profitability']])

least_profitable = df[df['Profitability'] == min_profit]
least_profitable
print(least_profitable[['Film', 'Profitability']])
```

```python
print(df['Genre'].unique())
df['Genre'] = df['Genre'].replace(
    {
        'Romence' : 'Romance',
        'Comdy' : 'Comedy',
        'romance' : 'Romance',
        'comedy' : 'Comedy',
    }
)
print(df['Genre'].unique())

```

```python
comedies = df[df['Genre'] == 'Comedy']
top_10_comedies = comedies.sort_values(by='Audience score %', ascending=False).head(10)
result = top_10_comedies[['Film', 'Year', 'Lead Studio']]
result.to_csv('top_10_comedies.csv', index=False)
print(result)
```
